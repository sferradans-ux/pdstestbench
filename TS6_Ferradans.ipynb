# TS6: ERespuesta en frecuencia de Filtros Pasivos RLC

## 1. Objetivo

En esta tarea se analizan tres circuitos RLC de segundo orden.

Para cada circuito se determinará analíticamente la función de transferencia

$$
H(s)=\frac{V_o(s)}{V_i(s)}
$$

y se calculará su respuesta en frecuencia para

$$
\omega_0=1\ \text{rad/s},
\qquad
R=1\ \Omega,
\qquad
Q=\sqrt{2}/2.
$$

Se estudiarán:

- respuesta de módulo;
- respuesta de fase;
- comportamiento asintótico;
- tipo de filtro;
- frecuencias de corte;
- ancho de banda.

Finalmente, como extensión, se calcularán nuevos valores de los componentes para que los filtros operen a

$$
f_0=10\ \text{kHz}
$$

utilizando

$$
R=50\ \Omega.
$$

## Circuito a)

El circuito está formado por una inductancia \(L\) y una resistencia \(R\) en serie, con un capacitor \(C\) conectado a masa. La salida \(V_o\) se toma sobre el capacitor.

La impedancia del capacitor es

$$
Z_C=\frac{1}{sC}.
$$

Aplicando divisor de tensión:

$$
H_a(s)=\frac{V_o(s)}{V_i(s)}
=
\frac{Z_C}{sL+R+Z_C}.
$$

Reemplazando \(Z_C\):

$$
H_a(s)
=
\frac{\frac{1}{sC}}
{sL+R+\frac{1}{sC}}.
$$

Multiplicando numerador y denominador por \(sC\):

$$
H_a(s)
=
\frac{1}
{LCs^2+RCs+1}.
$$

Dividiendo numerador y denominador por \(LC\):

$$
H_a(s)
=
\frac{\frac{1}{LC}}
{s^2+\frac{R}{L}s+\frac{1}{LC}}.
$$

Comparando con la forma general de un sistema de segundo orden:

$$
H(s)=
\frac{\omega_0^2}
{s^2+\frac{\omega_0}{Q}s+\omega_0^2},
$$

se obtiene:

$$
\omega_0^2=\frac{1}{LC}
$$

y

$$
\frac{\omega_0}{Q}=\frac{R}{L}.
$$

Por lo tanto:

$$
L=\frac{QR}{\omega_0}.
$$

Con los valores

$$
Q=\frac{\sqrt{2}}{2},
\qquad
R=1\ \Omega,
\qquad
\omega_0=1\ \text{rad/s},
$$

resulta:

$$
L=
\frac{\frac{\sqrt2}{2}\cdot1}{1}
=
\frac{\sqrt2}{2}\ \text{H}.
$$

Por lo tanto:

$$
\boxed{L\approx0.7071\ \text{H}}.
$$

A partir de

$$
\omega_0^2=\frac{1}{LC}
$$

se obtiene:

$$
C=\frac{1}{\omega_0^2L}.
$$

Entonces:

$$
C=
\frac{1}
{1^2\cdot\frac{\sqrt2}{2}}
=
\sqrt2\ \text{F}.
$$

Por lo tanto:

$$
\boxed{C\approx1.4142\ \text{F}}.
$$

La función de transferencia queda:

$$
\boxed{
H_a(s)=
\frac{1}
{s^2+\sqrt2\,s+1}
}.
$$

Como

$$
H_a(0)=1
$$

y

$$
H_a(j\omega)\rightarrow0
\qquad
\text{para }
\omega\rightarrow\infty,
$$

el circuito se comporta como un filtro pasabajos de segundo orden.

Para hallar la frecuencia de corte se parte de la condición de media potencia:

$$
|H(j\omega_c)|^2=\frac12.
$$

Reemplazando \(s=j\omega\):

$$
H_a(j\omega)
=
\frac{1}
{1-\omega^2+j\sqrt2\,\omega}.
$$

Por lo tanto:

$$
|H_a(j\omega)|^2
=
\frac{1}
{(1-\omega^2)^2+2\omega^2}.
$$

Como

$$
(1-\omega^2)^2+2\omega^2
=
1+\omega^4,
$$

queda:

$$
|H_a(j\omega)|^2
=
\frac{1}{1+\omega^4}.
$$

Imponiendo la condición de corte:

$$
\frac{1}{1+\omega_c^4}
=
\frac12.
$$

Entonces:

$$
2=1+\omega_c^4
$$

y por lo tanto:

$$
\omega_c^4=1.
$$

Tomando la solución positiva:

$$
\boxed{\omega_c=1\ \text{rad/s}}.
$$

En este caso:

$$
\boxed{\omega_c=\omega_0}.
$$

Esto ocurre porque

$$
Q=\frac{1}{\sqrt2},
$$

correspondiente a una respuesta Butterworth de segundo orden.

El ancho de banda del filtro pasabajos se toma desde continua hasta la frecuencia de corte, por lo que:

$$
\boxed{BW_a=1\ \text{rad/s}}.
$$

La respuesta asintótica de módulo es:

$$
|H_a|\approx1
\qquad
\text{para }
\omega\ll\omega_0,
$$

equivalente a:

$$
0\ \text{dB}.
$$

Para frecuencias altas:

$$
|H_a|\approx\frac{1}{\omega^2}
\qquad
\text{para }
\omega\gg\omega_0,
$$

por lo que la pendiente asintótica es:

$$
\boxed{-40\ \text{dB/década}}.
$$

## Circuito b)

En este circuito, la resistencia $R$ y el capacitor $C$ se encuentran conectados en paralelo entre la entrada y la salida, mientras que la inductancia $L$ se encuentra conectada entre la salida y masa.

La impedancia equivalente del paralelo $RC$ es:

$$
Z_{RC}=R\parallel\frac{1}{sC}
$$

por lo tanto:

$$
Z_{RC}
=
\frac{R\frac{1}{sC}}
{R+\frac{1}{sC}}
=
\frac{R}{1+sRC}
$$

La impedancia de la inductancia es:

$$
Z_L=sL
$$

Aplicando divisor de tensión:

$$
H_b(s)
=
\frac{V_o(s)}{V_i(s)}
=
\frac{Z_L}{Z_{RC}+Z_L}
$$

Reemplazando las impedancias:

$$
H_b(s)
=
\frac{sL}
{\frac{R}{1+sRC}+sL}
$$

Multiplicando numerador y denominador por $(1+sRC)$:

$$
H_b(s)
=
\frac{sL(1+sRC)}
{R+sL+s^2LRC}
$$

Desarrollando el numerador:

$$
H_b(s)
=
\frac{sL+s^2LRC}
{R+sL+s^2LRC}
$$

Dividiendo numerador y denominador por $LRC$:

$$
H_b(s)
=
\frac{s^2+\frac{1}{RC}s}
{s^2+\frac{1}{RC}s+\frac{1}{LC}}
$$

Comparando el denominador con la forma general de un sistema de segundo orden:

$$
s^2+\frac{\omega_0}{Q}s+\omega_0^2
$$

se obtiene:

$$
\frac{1}{RC}=\frac{\omega_0}{Q}
$$

y

$$
\frac{1}{LC}=\omega_0^2
$$

De la primera expresión:

$$
C=\frac{Q}{\omega_0R}
$$

Utilizando:

$$
Q=\frac{\sqrt2}{2},
\qquad
\omega_0=1\ \text{rad/s},
\qquad
R=1\ \Omega
$$

se obtiene:

$$
C
=
\frac{\frac{\sqrt2}{2}}
{1\cdot1}
=
\frac{\sqrt2}{2}\ \text{F}
$$

Por lo tanto:

$$
\boxed{C\approx0.7071\ \text{F}}
$$

A partir de:

$$
\frac{1}{LC}=\omega_0^2
$$

se despeja:

$$
L=\frac{1}{\omega_0^2C}
$$

y reemplazando:

$$
L
=
\frac{1}
{1^2\frac{\sqrt2}{2}}
=
\sqrt2\ \text{H}
$$

Por lo tanto:

$$
\boxed{L\approx1.4142\ \text{H}}
$$

Como:

$$
\frac{1}{RC}=\sqrt2
$$

y

$$
\frac{1}{LC}=1
$$

la función de transferencia queda:

$$
\boxed{
H_b(s)
=
\frac{s^2+\sqrt2\,s}
{s^2+\sqrt2\,s+1}
}
$$

### Tipo de filtro

Para frecuencia nula:

$$
H_b(0)=0
$$

mientras que para frecuencias muy altas:

$$
\lim_{\omega\rightarrow\infty}|H_b(j\omega)|=1
$$

Por lo tanto, el circuito presenta un comportamiento **pasaaltos**.

Cabe destacar que no corresponde al pasaaltos Butterworth canónico, ya que el numerador contiene tanto un término en $s^2$ como un término en $s$.

### Frecuencia de corte

Para calcular la frecuencia de corte se utiliza la condición de media potencia:

$$
|H_b(j\omega_c)|^2=\frac{1}{2}
$$

Reemplazando $s=j\omega$:

$$
H_b(j\omega)
=
\frac{(j\omega)^2+\sqrt2(j\omega)}
{(j\omega)^2+\sqrt2(j\omega)+1}
$$

Como:

$$
(j\omega)^2=-\omega^2
$$

resulta:

$$
H_b(j\omega)
=
\frac{-\omega^2+j\sqrt2\,\omega}
{1-\omega^2+j\sqrt2\,\omega}
$$

El módulo al cuadrado resulta:

$$
|H_b(j\omega)|^2
=
\frac{
(-\omega^2)^2+(\sqrt2\,\omega)^2
}{
(1-\omega^2)^2+(\sqrt2\,\omega)^2
}
$$

Por lo tanto:

$$
|H_b(j\omega)|^2
=
\frac{
\omega^4+2\omega^2
}{
(1-\omega^2)^2+2\omega^2
}
$$

Desarrollando el denominador:

$$
(1-\omega^2)^2+2\omega^2
=
1-2\omega^2+\omega^4+2\omega^2
$$

$$
=1+\omega^4
$$

Entonces:

$$
|H_b(j\omega)|^2
=
\frac{\omega^4+2\omega^2}
{1+\omega^4}
$$

Aplicando la condición de corte:

$$
\frac{\omega_c^4+2\omega_c^2}
{1+\omega_c^4}
=
\frac12
$$

Multiplicando ambos miembros:

$$
2\omega_c^4+4\omega_c^2
=
1+\omega_c^4
$$

Reordenando:

$$
\omega_c^4+4\omega_c^2-1=0
$$

Realizando el cambio:

$$
x=\omega_c^2
$$

queda:

$$
x^2+4x-1=0
$$

Aplicando la fórmula resolvente:

$$
x=
\frac{-4\pm\sqrt{4^2-4(1)(-1)}}{2}
$$

$$
x=
\frac{-4\pm\sqrt{20}}{2}
$$

$$
x=-2\pm\sqrt5
$$

Como $x=\omega_c^2$ debe ser positivo, se toma:

$$
\omega_c^2=\sqrt5-2
$$

Por lo tanto:

$$
\boxed{
\omega_c=\sqrt{\sqrt5-2}
}
$$

y numéricamente:

$$
\boxed{
\omega_c\approx0.486\ \text{rad/s}
}
$$

### Ancho de banda

Al tratarse de un filtro pasaaltos, su banda pasante comienza aproximadamente en:

$$
\omega_c=0.486\ \text{rad/s}
$$

y se extiende idealmente hasta frecuencia infinita. Por lo tanto, no posee un ancho de banda superior finito.

### Respuesta asintótica

Para frecuencias mucho menores que $\omega_0$, el término dominante del numerador es $\sqrt2\,s$, mientras que el denominador tiende a $1$:

$$
H_b(j\omega)\approx j\sqrt2\,\omega
$$

por lo tanto:

$$
|H_b(j\omega)|\approx\sqrt2\,\omega
$$

lo que corresponde a una pendiente asintótica de:

$$
\boxed{+20\ \text{dB/década}}
$$

Para frecuencias mucho mayores que $\omega_0$, los términos en $s^2$ dominan tanto en el numerador como en el denominador:

$$
H_b(j\omega)\approx
\frac{(j\omega)^2}{(j\omega)^2}=1
$$

por lo tanto, la respuesta asintótica tiende a:

$$
\boxed{0\ \text{dB}}
$$

## Circuito c)

En este circuito, la resistencia $R$ se encuentra en serie con una asociación en paralelo formada por la inductancia $L$ y el capacitor $C$. La salida $V_o$ se toma sobre la rama paralela $LC$.

La impedancia de la inductancia es:

$$
Z_L=sL
$$

y la impedancia del capacitor es:

$$
Z_C=\frac{1}{sC}
$$

La impedancia equivalente del paralelo $LC$ es:

$$
Z_{LC}=Z_L\parallel Z_C
$$

por lo tanto:

$$
Z_{LC}
=
\frac{Z_L Z_C}{Z_L+Z_C}
$$

Reemplazando:

$$
Z_{LC}
=
\frac{sL\frac{1}{sC}}
{sL+\frac{1}{sC}}
$$

Multiplicando numerador y denominador por $sC$:

$$
Z_{LC}
=
\frac{sL}
{1+s^2LC}
$$

Aplicando divisor de tensión:

$$
H_c(s)
=
\frac{V_o(s)}{V_i(s)}
=
\frac{Z_{LC}}
{R+Z_{LC}}
$$

Reemplazando $Z_{LC}$:

$$
H_c(s)
=
\frac{\frac{sL}{1+s^2LC}}
{R+\frac{sL}{1+s^2LC}}
$$

Multiplicando numerador y denominador por $(1+s^2LC)$:

$$
H_c(s)
=
\frac{sL}
{R+sL+s^2RLC}
$$

Dividiendo numerador y denominador por $RLC$:

$$
H_c(s)
=
\frac{\frac{1}{RC}s}
{s^2+\frac{1}{RC}s+\frac{1}{LC}}
$$

Comparando el denominador con la forma general de un sistema de segundo orden:

$$
s^2+\frac{\omega_0}{Q}s+\omega_0^2
$$

se obtiene:

$$
\frac{1}{RC}=\frac{\omega_0}{Q}
$$

y

$$
\frac{1}{LC}=\omega_0^2
$$

De la primera expresión:

$$
C=\frac{Q}{\omega_0R}
$$

Usando:

$$
Q=\frac{\sqrt2}{2},
\qquad
\omega_0=1\ \text{rad/s},
\qquad
R=1\ \Omega
$$

resulta:

$$
C=
\frac{\frac{\sqrt2}{2}}
{1\cdot1}
=
\frac{\sqrt2}{2}\ \text{F}
$$

Por lo tanto:

$$
\boxed{C\approx0.7071\ \text{F}}
$$

A partir de:

$$
\frac{1}{LC}=\omega_0^2
$$

se despeja:

$$
L=\frac{1}{\omega_0^2C}
$$

Reemplazando:

$$
L=
\frac{1}
{1^2\frac{\sqrt2}{2}}
=
\sqrt2\ \text{H}
$$

Por lo tanto:

$$
\boxed{L\approx1.4142\ \text{H}}
$$

Como:

$$
\frac{1}{RC}=\sqrt2
$$

y

$$
\frac{1}{LC}=1
$$

la función de transferencia queda:

$$
\boxed{
H_c(s)
=
\frac{\sqrt2\,s}
{s^2+\sqrt2\,s+1}
}
$$

### Tipo de filtro

Para frecuencia nula:

$$
H_c(0)=0
$$

y para frecuencias muy altas:

$$
\lim_{\omega\rightarrow\infty}|H_c(j\omega)|=0
$$

En cambio, la respuesta alcanza su máximo alrededor de:

$$
\omega_0=1\ \text{rad/s}
$$

Por lo tanto, el circuito se comporta como un filtro **pasabanda de segundo orden**.

### Respuesta en la frecuencia central

Reemplazando:

$$
s=j\omega
$$

se obtiene:

$$
H_c(j\omega)
=
\frac{j\sqrt2\,\omega}
{1-\omega^2+j\sqrt2\,\omega}
$$

Para:

$$
\omega=\omega_0=1
$$

queda:

$$
H_c(j)
=
\frac{j\sqrt2}
{j\sqrt2}
=1
$$

por lo tanto:

$$
\boxed{|H_c(j\omega_0)|=1}
$$

equivalente a:

$$
\boxed{0\ \text{dB}}
$$

### Frecuencias de corte

Las frecuencias de corte se obtienen a partir de la condición de media potencia:

$$
|H_c(j\omega)|^2=\frac12
$$

El módulo al cuadrado es:

$$
|H_c(j\omega)|^2
=
\frac{(\sqrt2\,\omega)^2}
{(1-\omega^2)^2+(\sqrt2\,\omega)^2}
$$

Por lo tanto:

$$
|H_c(j\omega)|^2
=
\frac{2\omega^2}
{(1-\omega^2)^2+2\omega^2}
$$

Desarrollando el denominador:

$$
(1-\omega^2)^2+2\omega^2
=
1-2\omega^2+\omega^4+2\omega^2
$$

$$
=1+\omega^4
$$

Entonces:

$$
|H_c(j\omega)|^2
=
\frac{2\omega^2}
{1+\omega^4}
$$

Aplicando la condición de corte:

$$
\frac{2\omega^2}
{1+\omega^4}
=
\frac12
$$

Multiplicando:

$$
4\omega^2=1+\omega^4
$$

Reordenando:

$$
\omega^4-4\omega^2+1=0
$$

Realizando el cambio:

$$
x=\omega^2
$$

queda:

$$
x^2-4x+1=0
$$

Aplicando la fórmula resolvente:

$$
x=
\frac{4\pm\sqrt{16-4}}{2}
$$

$$
x=
\frac{4\pm\sqrt{12}}{2}
$$

$$
x=2\pm\sqrt3
$$

Como:

$$
x=\omega^2
$$

se obtienen las dos frecuencias de corte:

$$
\boxed{
\omega_L=\sqrt{2-\sqrt3}
}
$$

y:

$$
\boxed{
\omega_H=\sqrt{2+\sqrt3}
}
$$

Numéricamente:

$$
\boxed{
\omega_L\approx0.518\ \text{rad/s}
}
$$

$$
\boxed{
\omega_H\approx1.932\ \text{rad/s}
}
$$

### Ancho de banda

El ancho de banda se define como:

$$
BW=\omega_H-\omega_L
$$

por lo tanto:

$$
BW
=
1.932-0.518
$$

$$
\boxed{
BW\approx1.414\ \text{rad/s}
}
$$

Este resultado coincide con la relación teórica:

$$
BW=\frac{\omega_0}{Q}
$$

ya que:

$$
BW=
\frac{1}{\frac{\sqrt2}{2}}
=
\sqrt2
$$

por lo tanto:

$$
\boxed{
BW=\sqrt2\approx1.414\ \text{rad/s}
}
$$

Además, se verifica la relación:

$$
\omega_0=\sqrt{\omega_L\omega_H}
$$

ya que:

$$
\sqrt{0.518\cdot1.932}\approx1
$$

### Respuesta asintótica

Para frecuencias mucho menores que $\omega_0$:

$$
\omega\ll\omega_0
$$

el denominador tiende a $1$, por lo que:

$$
H_c(j\omega)\approx j\sqrt2\,\omega
$$

y:

$$
|H_c(j\omega)|\approx\sqrt2\,\omega
$$

Esto corresponde a una pendiente asintótica de:

$$
\boxed{+20\ \text{dB/década}}
$$

Para frecuencias mucho mayores que $\omega_0$, domina el término $s^2$ del denominador:

$$
H_c(j\omega)
\approx
\frac{j\sqrt2\,\omega}
{(j\omega)^2}
$$

por lo que:

$$
|H_c(j\omega)|
\approx
\frac{\sqrt2}{\omega}
$$

y la pendiente asintótica resulta:

$$
\boxed{-20\ \text{dB/década}}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

R = 1
w0 = 1
Q = 1/np.sqrt(2)

den = [1, w0/Q, w0**2]

# a) pasabajos
num_a = [w0**2]

# b) pasaaltos
num_b = [1, w0/Q, 0]

# c) pasabanda
num_c = [w0/Q, 0]

Ha = signal.TransferFunction(num_a, den)
Hb = signal.TransferFunction(num_b, den)
Hc = signal.TransferFunction(num_c, den)

w = np.logspace(-2, 2, 2000)
```

```python
def bode_plot(sys, nombre, w):

    w_resp, H = signal.freqresp(sys, w)

    modulo_db = 20*np.log10(np.abs(H))
    fase = np.unwrap(np.angle(H))*180/np.pi

    plt.figure(figsize=(10, 5))
    plt.semilogx(w_resp, modulo_db)
    plt.xlabel(r'Frecuencia angular $\omega$ [rad/s]')
    plt.ylabel('Módulo [dB]')
    plt.title(nombre + ' — módulo')
    plt.grid(True, which='both')
    plt.show()

    plt.figure(figsize=(10, 5))
    plt.semilogx(w_resp, fase)
    plt.xlabel(r'Frecuencia angular $\omega$ [rad/s]')
    plt.ylabel('Fase [°]')
    plt.title(nombre + ' — fase')
    plt.grid(True, which='both')
    plt.show()
```

```python
bode_plot(Ha, 'Circuito a — Pasabajos', w)
bode_plot(Hb, 'Circuito b — Pasaaltos', w)
bode_plot(Hc, 'Circuito c — Pasabanda', w)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Parámetros
w0 = 1
Q = 1/np.sqrt(2)

# Vector de frecuencias
w = np.logspace(-2, 2, 2000)

# Denominador común
den = [1, w0/Q, w0**2]

# Funciones de transferencia
Ha = signal.TransferFunction(
    [w0**2],
    den
)

Hb = signal.TransferFunction(
    [1, w0/Q, 0],
    den
)

Hc = signal.TransferFunction(
    [w0/Q, 0],
    den
)

# Respuestas exactas
_, Ha_resp = signal.freqresp(Ha, w)
_, Hb_resp = signal.freqresp(Hb, w)
_, Hc_resp = signal.freqresp(Hc, w)

Ha_db = 20*np.log10(np.abs(Ha_resp))
Hb_db = 20*np.log10(np.abs(Hb_resp))
Hc_db = 20*np.log10(np.abs(Hc_resp))
```

```python
# ---------------------------
# Asíntotas
# ---------------------------

# a) Pasabajos
asintota_a = np.where(
    w <= w0,
    0,
    -40*np.log10(w/w0)
)

# b) Pasaaltos
asintota_b = np.where(
    w <= w0,
    20*np.log10(np.sqrt(2)*w),
    0
)

# c) Pasabanda
asintota_c = np.where(
    w <= w0,
    20*np.log10(np.sqrt(2)*w),
    20*np.log10(np.sqrt(2)/w)
)
```

```python
plt.figure(figsize=(10, 5))

plt.semilogx(
    w,
    Ha_db,
    label='Respuesta exacta'
)

plt.semilogx(
    w,
    asintota_a,
    '--',
    label='Respuesta asintótica'
)

plt.axvline(
    w0,
    linestyle=':',
    label=r'$\omega_0 = 1$ rad/s'
)

plt.xlabel(r'Frecuencia angular $\omega$ [rad/s]')
plt.ylabel('Módulo [dB]')
plt.title('Circuito a — Filtro pasabajos')
plt.legend()
plt.grid(True, which='both')
plt.show()
```

## Análisis de los resultados

### Circuito a)

El circuito a presenta comportamiento pasabajos de segundo orden.

Para bajas frecuencias la ganancia tiende a:

$$
|H_a|\approx1
$$

equivalente a:

$$
0\ \text{dB}
$$

A partir de la frecuencia característica la respuesta comienza a decrecer y, para frecuencias elevadas, presenta una pendiente asintótica de:

$$
-40\ \text{dB/década}
$$

El valor:

$$
Q=\frac{1}{\sqrt2}
$$

corresponde al caso Butterworth de segundo orden, por lo que la respuesta de módulo es máximamente plana en la banda pasante y la frecuencia de corte coincide con la frecuencia natural:

$$
\omega_c=\omega_0=1\ \text{rad/s}
$$

En la frecuencia de corte la potencia se reduce a la mitad respecto del valor máximo, equivalente a una caída de aproximadamente $3$ dB.

### Circuito b)

El circuito b presenta comportamiento pasaaltos.

A muy bajas frecuencias la salida tiende a cero, mientras que para frecuencias elevadas la ganancia tiende a:

$$
|H_b|\rightarrow1
$$

equivalente a:

$$
0\ \text{dB}
$$

La respuesta asintótica a bajas frecuencias presenta una pendiente aproximada de:

$$
+20\ \text{dB/década}
$$

A diferencia de un pasaaltos Butterworth canónico, el numerador de este circuito contiene un término en $s$ además del término en $s^2$:

$$
s^2+\sqrt2\,s
$$

Por esta razón, aunque el denominador tenga la forma Butterworth, la frecuencia de corte de la función completa no coincide con $\omega_0$.

Se obtuvo:

$$
\omega_c\approx0.486\ \text{rad/s}
$$

La banda pasante se extiende desde dicha frecuencia hacia frecuencias arbitrariamente altas, por lo que idealmente no posee un ancho de banda superior finito.

### Circuito c)

El circuito c presenta comportamiento pasabanda.

Para bajas frecuencias la ganancia aumenta aproximadamente con una pendiente de:

$$
+20\ \text{dB/década}
$$

mientras que para frecuencias superiores a la frecuencia central disminuye con una pendiente aproximada de:

$$
-20\ \text{dB/década}
$$

La ganancia máxima se produce en:

$$
\omega_0=1\ \text{rad/s}
$$

y su valor es:

$$
|H_c(j\omega_0)|=1
$$

equivalente a:

$$
0\ \text{dB}
$$

Las frecuencias de corte obtenidas fueron:

$$
\omega_L\approx0.518\ \text{rad/s}
$$

y:

$$
\omega_H\approx1.932\ \text{rad/s}
$$

por lo que el ancho de banda es:

$$
BW=\omega_H-\omega_L
$$

$$
BW\approx1.414\ \text{rad/s}
$$

Este resultado coincide con la relación teórica:

$$
BW=\frac{\omega_0}{Q}
$$

ya que:

$$
\frac{1}{\sqrt2/2}=\sqrt2\approx1.414
$$

El análisis de los tres circuitos permitió observar que diferentes topologías construidas con los mismos tipos de componentes pueden producir respuestas en frecuencia muy distintas.

El circuito a se comporta como un filtro pasabajos de segundo orden. Debido a que:

$$

su respuesta corresponde al caso Butterworth, por lo que la frecuencia de corte coincide con la frecuencia natural:

$$
\omega_c\approx0.486\ \text{rad/s}
$$

y no coincide con $\omega_0$.

El circuito c presenta comportamiento pasabanda, con máxima transferencia en:

$$
\omega_0=1\ \text{rad/s}
$$

y un ancho de banda de:

$$
BW\approx1.414\ \text{rad/s}
$$

verificándose la relación:

$$
BW=\frac{\omega_0}{Q}
$$

Los resultados obtenidos analíticamente son consistentes con las respuestas de módulo y fase observadas en los gráficos.

En conjunto, los tres casos muestran cómo tanto la ubicación de los componentes reactivos como la forma del numerador de la función de transferencia determinan el comportamiento espectral final del circuito.
